## Set-up the project environment for CircuitBox Data Lakehouse

- 1. Create external location - dea_course_ext_dl_circuitbox
- 2. Create Catalog - circuitbox
- 3. Create Schemas
    - landing
    - lakehouse
- 4. Create Volume - operational_data


![CircuitBox_setup_pic.png](./CircuitBox_setup_pic.png "CircuitBox_setup_pic.png")


1. Create external Location - dea_course_ext_dl_circuitbox

```sql
CREATE EXTERNAL LOCATION IF NOT EXISTS dea_course_ext_dl_circuitbox
    URL 'abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/'
    WITH (STORAGE CREDENTIAL  dbxcoursestoragecredential)
    COMMENT 'External Location for the circuitbox data lakehouse'
```

```sql
%fs ls 'abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/'
```

### 2. Create catalog
- Catalog name: circuitbox
- managed location: abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/

```sql
CREATE CATALOG IF NOT EXISTS circuitbox
    MANAGED LOCATION 'abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/'
    COMMENT 'Catalog for the circuitbox data lakehouse'
```

```sql
SHOW CATALOGS;
```

### 3. Create Schemas

1. Schema Name: landing
    Managed Location: 'abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/'
2. Schema Name: lakehouse
    Managed Location: 'abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/'

```sql
USE CATALOG circuitbox;

CREATE SCHEMA IF NOT EXISTS landing
    MANAGED LOCATION 'abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/landing';

CREATE SCHEMA IF NOT EXISTS lakehouse
    MANAGED LOCATION 'abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/lakehouse';

```

```sql
SHOW SCHEMAS;
```

### 4. Create Volume
- Volume Name: operational_data
- ADLS Path: abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/landing/operational_data/

```sql
USE CATALOG circuitbox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
    LOCATION 'abfss://circuitbox@dbxextstorageaccount.dfs.core.windows.net/landing/operational_data'
    COMMENT 'Volume for the circuitbox landing operational_data';
```

```sql
SHOW VOLUMES;
```
